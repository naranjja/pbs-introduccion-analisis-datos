```python
# Concatenación
print([1, 2, 3] + [99, "hola"])
print((1, 2) + (3, "chau"))
print("ABC" + "DEF")

# f-strings (preferido)
nombre = "jose"
print(f"mi nombre es {nombre}")

num_perros = 1
print(f"hay {num_perros} perro{'' if num_perros == 1 else 's'}")

# Casting numérico
print(float(10))
print(int(3.145646))       # trunca, no redondea
print(int(round(99.9, 0)))

# Aritméticos
n1, n2 = 1, 4
print(n1 + n2, n1 - n2, n1 * n2, n1 / n2, n1 ** n2)
print(100 // 3)   # división entera
print(12 % 5)     # módulo (residuo)

# Par / impar
x = 1
print("par" if x % 2 == 0 else "impar")

# Asignación aumentada (no existe x++)
x = 1
x += 3
x -= 1
x *= 10
print(x)

# Asignación múltiple
nombre, apellido, edad = "jose", "naranjo", 32
print(nombre, apellido, edad)
```

<div class="pbs-section">
  <span class="num">03</span>
  <h2>Manejo de memoria</h2>
</div>

```python
x = 100
y = 100
print(x == y)   # mismo valor
print(x is y)   # ¿misma referencia? (enteros pequeños suelen compartirse)

l1 = [1, 2, 3]
l2 = [1, 2, 3]
print(l1 == l2)   # True  - mismo contenido
print(l1 is l2)   # False - objetos distintos
print(id(l1), id(l2))

# ¡Cuidado con referencias!
a = [1, 2, 3]
b = a
b.append(4)
print("a", a, "b", b)  # ambas cambian

a = [1, 2, 3]
b = a.copy()
b.append(4)
print("a", a, "b", b)  # solo b cambia
```

<div class="pbs-section light">
  <span class="num">04</span>
  <h2>Manejo de errores (EAFP)</h2>
</div>

```python
a, b = 3, 0
try:
    print(a / b)
except ZeroDivisionError:
    print("división entre cero no permitida")
except TypeError:
    print("has mezclado tipos incompatibles")
except Exception:
    print("error desconocido")

# Preferible intentar y capturar que validar todo de antemano
edad = 20
try:
    x = edad + 1
    x /= edad
    x += 100
    print(x)
except ZeroDivisionError:
    print("error, edad inválida")
```

```python
formato_correo = r"\S+@\S+\.\S+"
correo = "naranjja@gmail.com"
print("correo válido" if re.match(formato_correo, correo) else "correo inválido")

formato_telf = r"\+?[1-9][0-9]{7,14}"
telf = "+51981315163"
if re.match(formato_telf, telf):
    print("teléfono válido")

msg = "hola, mi correo es naranjja@gmail.com o abc@ppp.com y mi teléfono el +51981315163"
print(re.findall(formato_correo, msg))
print(re.findall(formato_telf, msg))
```

  <h2>Clases (intro OOP)</h2>
</div>

```python
class Koopa:
    def __init__(self, caparazon, nombre):
        self.caparazon = caparazon
        self.nombre = nombre

    def saludar(self):
        if self.caparazon:
            print(f"Hola me llamo {self.nombre} y tengo caparazón {self.caparazon}!")
        else:
            print(f"Hola me llamo {self.nombre} y no tengo caparazón!")


class Magikoopa(Koopa):
    def __init__(self, nombre):
        super().__init__(None, nombre)


Koopa("verde", "Luis").saludar()
Magikoopa("Kamek").saludar()
```
